```python
import boto3
import pandas as pd
import io

# --- CẤU HÌNH MINIO ---
minio_config = {
    "endpoint_url": "http://localhost:9000",
    "aws_access_key_id": "minio_admin",
    "aws_secret_access_key": "minio_secret_secure_password_123",
    "region_name": "us-east-1"
}

# Tên bucket và Key (đường dẫn object) bạn đã tìm thấy
bucket_name = "bronze-lake"
object_key = "date=2020-01-02/batch_data.parquet"

print(f"🔍 Đang tải object '{object_key}' về RAM...")

s3_client = boto3.client('s3', **minio_config)

try:
    # 1. Tải nội dung file vào bộ nhớ đệm (BytesIO)
    response = s3_client.get_object(Bucket=bucket_name, Key=object_key)
    # Đọc body stream
    file_content = response['Body'].read()
    
    # Đưa vào buffer
    buffer = io.BytesIO(file_content)
    print("✅ Đã tải xong! Kích thước:", len(file_content), "bytes")

    # 2. Đọc Parquet từ buffer (Pandas sẽ không biết đường dẫn nữa -> Không bị lỗi merge date)
    print("🚀 Đang đọc DataFrame từ buffer...")
    df = pd.read_parquet(buffer)
    
    print("\n🎉 ĐỌC THÀNH CÔNG RỰC RỠ!")
    print("-" * 30)
    print(f"Số dòng: {len(df)}")
    print(f"Các cột: {list(df.columns)}")
    print("-" * 30)
    print(df.head())

except Exception as e:
    print(f"\n❌ LỖI: {e}")
```
