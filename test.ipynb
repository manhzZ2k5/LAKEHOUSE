```python
import pandas as pd
import s3fs
```

```python
# --- 1. KHAI BÁO CẤU HÌNH (QUAN TRỌNG) ---
minio_config = {
    "key": "minio_admin",                        # User
    "secret": "minio_secret_secure_password_123", # Password
    "client_kwargs": {
        "endpoint_url": "http://localhost:9000"   # Cổng API của MinIO (localhost vì chạy notebook ngoài docker)
    }
}

# --- 2. ĐƯỜNG DẪN FILE ---
# Lưu ý: 
# - Phải có s3:// ở đầu
# - bucket là 'bronze-lake' (không phải 'data_lake/bronze-lake' vì đó là đường dẫn folder mount)
# - Thêm folder 'date=2020-01-01' vào đường dẫn
path = "s3://bronze-lake/date=2020-01-01/batch_data.parquet"

print(f"📡 Đang kết nối thử đến: {path} ...")

# --- 3. TIẾN HÀNH ĐỌC ---
try:
    # Đọc file trực tiếp từ MinIO
    df = pd.read_parquet(
        path, 
        storage_options=minio_config  # Truyền cấu hình vào đây
    )
    
    print("\n🎉 THÀNH CÔNG RỰC RỠ!")
    print("-" * 30)
    print(f"Số dòng dữ liệu: {len(df)}")
    print(f"Số cột dữ liệu: {len(df.columns)}")
    print("-" * 30)
    print("5 dòng đầu tiên:")
    print(df.head()) 

except Exception as e:
    print("\n❌ VẪN CÒN LỖI:")
    print(e)
    print("\n👉 Gợi ý kiểm tra:")
    print("1. Kiểm tra xem Docker có đang chạy không (docker ps)")
    print("2. Kiểm tra xem bucket 'bronze-lake' và file có thật sự tồn tại trên http://localhost:9001 không")
```
